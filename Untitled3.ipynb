```python
import cfgrib

# Read the GRIB file into an xarray Dataset
grib_file = 'download_march_2024.grib'
ds = cfgrib.open_dataset(grib_file)

# Print the Dataset information
print(ds)

# Access specific variables (e.g., geopotential)
geopotential = ds['z']

# Print the geopotential data
print(geopotential)

# You can also access specific values in the variable
print(geopotential.sel(time='2024-03-01T00:00:00'))

# Close the dataset when done
ds.close()
```
